```python
import pandas as pd
import matplotlib.pyplot as plt
import requests
import sqlite3
```

```python
api_url = "https://api.github.com/search/repositories?q=machine+learning&sort=stars&order=desc&per_page=100"
response= requests.get(api_url)
data = response.json()
```

```python
df = pd.DataFrame(data["items"])
```

```python
df.head()
```

```python
df.columns
```

```python
# Extract the owner login name
df["owner"] = df["owner"].apply(lambda x: x["login"] if type(x) == dict else x)

# Extract the license name
df["license"] = df["license"].apply(
    lambda x: x["name"] if type(x) == dict else x
)
```

```python
# Replace missing values in 'language' and 'license' with 'Unknown'
df["language"] = df["language"].fillna("Unknown")
df["license"] = df["license"].fillna("Unknown")

# Check if there are any remaining missing values in the dataset
print(df.isnull().sum())
```

```python
df['topics'] = df['topics'].apply(str)

# Check how many duplicate rows exist
print("Duplicate rows before:", df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Check total rows left after cleaning:
print("Total rows after cleaning:", len(df))
```

```python
# Convert date columns to standard date format
df["created_at"] = pd.to_datetime(df["created_at"]).dt.date
df["updated_at"] = pd.to_datetime(df["updated_at"]).dt.date

# Rename columns to shorter names
df = df.rename(
    columns={
        "stargazers_count": "stars",
        "forks_count": "forks",
        "watchers_count": "watchers",
        "open_issues_count": "open_issues",
        "created_at": "created_date",
        "updated_at": "updated_date",
    }
)


df.head()
```

```python
#1 Save the cleaned DataFrame to a CSV file
df.to_csv("github_projects.csv", index=False)

# 2. Reload the CSV file
df_verify = pd.read_csv("github_projects.csv")

# 3 show the first rows
print(df_verify.head())

# 4. Display structure
print(df_verify.info())
```

```python
import pandas as pd
import sqlite3

conn = sqlite3.connect("github_data.db")

# Repositories with more than 10,000 stars
query1 = """
SELECT name, stars, language
FROM Repositories
WHERE stars > 10000
"""

df1 = pd.read_sql_query(query1, conn)
print(df1.head())


# Repositories with "Machine" in the name
query2 = """
SELECT name, owner, stars
FROM Repositories
WHERE name LIKE '%Machine%'
"""

df2 = pd.read_sql_query(query2, conn)
print(df2.head())

conn.close()
```

```python
import pandas as pd
import sqlite3

conn = sqlite3.connect("github_data.db")

# Python projects with more than 5000 stars
query1 = """
SELECT name, language, stars, license
FROM Repositories
WHERE language = 'Python'
AND stars > 5000
AND license != 'Unknown'
"""

df1 = pd.read_sql_query(query1, conn)
print(df1.head())


# Python or C++ projects
query2 = """
SELECT name, language, stars
FROM Repositories
WHERE language = 'Python'
OR language = 'C++'
"""

df2 = pd.read_sql_query(query2, conn)
print(df2.head())

conn.close()
```

```python
conn = sqlite3.connect("github_data.db")

query = """
SELECT name, owner, stars, language
FROM Repositories
ORDER BY stars DESC
LIMIT 10
"""

df = pd.read_sql_query(query, conn)

print(df)

conn.close()
```

```python
conn = sqlite3.connect("github_data.db")

query = """
SELECT COUNT(*) AS total_repositories,
       AVG(stars) AS average_stars
FROM Repositories
"""

df = pd.read_sql_query(query, conn)

print(df)

conn.close()
```

```python
import pandas as pd
import sqlite3

conn = sqlite3.connect("github_data.db")

query = """
SELECT language, COUNT(*) AS repository_count
FROM Repositories
GROUP BY language
HAVING COUNT(*) > 5
ORDER BY repository_count DESC
"""

df = pd.read_sql_query(query, conn)

print(df)

conn.close()
```

```python
conn = sqlite3.connect("github_data.db")

query = """
SELECT name, stars
FROM Repositories
ORDER BY stars DESC
LIMIT 10
"""

df = pd.read_sql_query(query, conn)

plt.barh(df["name"], df["stars"])
plt.xlabel("Stars")
plt.ylabel("Repository")
plt.title("Top 10 Repositories")
plt.show()

conn.close()
```

```python
conn = sqlite3.connect("github_data.db")

query = """
SELECT created_date, COUNT(*) AS count
FROM Repositories
GROUP BY created_date
ORDER BY created_date
"""

df = pd.read_sql_query(query, conn)

plt.plot(df["created_date"], df["count"])
plt.xlabel("Date")
plt.ylabel("Repositories")
plt.title("Repositories Created Over Time")
plt.xticks(rotation=45)
plt.show()

conn.close()
```

# **(Why is it important to verify data collected from public APIs?)**

Because Public APIs may contain outdated, missing, or incorrect data due to technical issues during data transmission. Data validation ensures the accuracy of the analysis and supports sound, reliable decision-making based on accurate sources.

# **2nd question:Why should data analysts document the source of their data?*

Because Documentation improves transparency and credibility, allowing others to verify the results and reproduce the analysis . It also protects intellectual property rights and clarifies the ethical and legal responsibilities related to data usage.

# **3rd question: (How can missing or inaccurate data affect data analysis and decision-making?)**


They may lead to biased results or completely incorrect conclusions, which can result in poor strategic decisions, wasted resources, or even project failure.
